$B(v)$ é dado pela quantidade de caminhos mínimos de vértices $i,j \neq v; \ i \neq j$ que passam por v sobre a quantidade total de arestas caminhos entre vértices $i,j; \ i \neq j$.

Para calcular $B(v)$ denotaremos por $S$ o conjunto de caminhos mínimos que passa por $v$ e por $N$ o conjunto de caminhos mínimos que não passa por $v$, sendo assim $B(v) = \frac{|S|}{|S|+|N|}$.

Como os únicos vértices ligados a $A$ são $B$ e $C$, e existe uma aresta direta $(B,C)$ sabemos que a centralidade de intermediação de $A$ é zero. Pela simetria do grafo o mesmo vale para $B, E \ \text{e} \ F$, portanto,  $B(A)=B(B)=B(E)=B(F)=0$

Para $C$ temos: $S = \{(AP),(AD),(AE),(AF),(BP),(BD),(BE),(BF)\}$ e $N = \{(AB),(PD),(PE),(PF),(DE),(DF),(EF)\}$

Como $|S| = 8$ e $|N| = 7$ temos $B(C) = \frac{8}{8+7} = \frac{8}{15}$ e por simetria $B(D) = B(C) = \frac{8}{15}$

Para $P$ temos: $S = \{(AD),(AE),(AF),(BD),(BE),(BF),(CD),(CE),(CF)\}$ e $N = \{(AB),(AC),(BC),(DE),(DF),(EF)$, portanto podemos calcular $B(P) = \frac{|S|}{|S|+|N|} = \frac{9}{15} = \frac{3}{5}$
