# Questão 2

Como a probabilidade de existir uma aresta para qualquer outro nó é $p = \frac{c}{N}$ sabemos que a esperança será dada por $E = \sum_{i=1}^N{\frac{c}{N}} = N\frac{c}{N} = c$

Seja $V$ a quantidade de nós já visitados então podemos calcular a esperança da quantidade de nós que ainda não foram visitados como $E - V = c - V$. Como $V$ é pequeno temos que $E \approx c$

Queremos mostrar que $\sum_{d=0}^{\infty}{c^d} = \frac{1}{1-c}$ para $0 \lt c \lt 1$.

Defina a soma até o $n$-ésimo termo como $S_n = \sum_{d=0}^n{c^d} = 1 + c + c^2 + \dots + c^n$

Ao multiplicar os dois lados por $c$ obtemos $c \cdot S_N = c + c^2 + c^3 + \dots + c^{N+1}$

Ao subtrair a primeira equação da segunda obtemos $S_N - c \cdot S_N = S_n (1-c) = (1 + c + c^2 + \dots + c^N) - (c + c^2 + \dots + c^{N+1}) = 1 - c^{N+1}$

Como $1-c \neq 0$ podemos dividir os dois lados por $1-c$ obtendo $S_n = \frac{1 - c^{N+1}}{1-c}$

Portanto só precisamos tirar o limite de $n$ no infinito. Como $\lim_{d \to \infty}{c^{d+1}} = 0 \Rightarrow \lim_{d \to \infty}{S_n} = \frac{1}{1-c}$

O passo essencial para mostrar que $E(|C(V)|)$ convergia era tomar $\lim_{d \to \infty}{c^d} = 0$. 

Quando $c > 1$ acontece o contrário, temos $\lim_{n \to \infty}{S_n} = \lim_{n \to \infty}{\frac{c^n - 1}{c-1}} = \infty$

Isso significa que quando $c > 1$ a maior componente conexa não é mais um subgrafo de tamanho limitado e sim um subgrafo de tamanho proporcional à $N$. O que é exatamente a condição necessária para a componente gigante.

Essa soma deixa de fazer sentido pois $\alpha \infty = \infty$ quando $\alpha > 0$ então a soma deixa de demonstrar o tamanho real da componente gigante e apenas mostra que o tamanho da componente gigante é proporcional à $N$

Quando $c < 1$ a proporção do tamanho da maior compontente conexa é $\alpha = \frac{\frac{1}{1-c}}{N}$, como $\frac{1}{1-c}$ não cresce junto com $N$ temos: $$\lim_{N \to \infty}{\alpha} = \lim_{N \to \infty}{\frac{\frac{1}{1-c}}{N}} = 0$$

Logo não pode existir uma componente gigante (componente com tamanho proporcional à $N$) nesse regime.
